Q1.A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

Dataset: student_assessment_dirty.csv

| Student_ID | Student_Name | Department | Assessment |    Marks |
| ---------- | ------------ | ---------- | ---------- | -------: |
| S101       | Asha         | IT         | Python     |       85 |
| S102       | Ravi         | it         | Python     |       78 |
| S103       | Neha         | IT         | Python     |       92 |
| S104       | Karan        | HR         | Python     |    blank |
| S105       | Meera        | HR         | Python     |   Absent |
| S106       | Arjun        | Finance    | Python     |        0 |
| S107       | Priya        | finance    | Python     |       67 |
| S108       | Rahul        | IT         | Python     |       75 |
| S102       | Ravi         | it         | Python     |       78 |
| S109       | Sneha        | I.T.       | Python     |       NA |
| S110       | Varun        | HR         | Python     | 88 marks |
| S111       | Diya         | Finance    | Python     |       91 |
| S112       | Aman         | IT         | Python     |    blank |
| S113       | Nisha        | HR         | Python     |       64 |
| S114       | Kabir        | Finance    | Python     |      105 |


```python
#Code
import pandas as pd
from io import StringIO

data = """Student_ID,Student_Name,Department,Assessment,Marks
S101,Asha,IT,Python,85
S102,Ravi,it,Python,78
S103,Neha,IT,Python,92
S104,Karan,HR,Python,
S105,Meera,HR,Python,Absent
S106,Arjun,Finance,Python,0
S107,Priya,finance,Python,67
S108,Rahul,IT,Python,75
S102,Ravi,it,Python,78
S109,Sneha,I.T.,Python,NA
S110,Varun,HR,Python,88 marks
S111,Diya,Finance,Python,91
S112,Aman,IT,Python,
S113,Nisha,HR,Python,64
S114,Kabir,Finance,Python,105"""

df = pd.read_csv(StringIO(data))

print(df.head())
print(df.info())
print(df.isnull().sum())

df["Department"] = df["Department"].str.strip().str.lower()

df["Department"] = df["Department"].replace({
    "it": "IT",
    "i.t.": "IT",
    "hr": "HR",
    "finance": "Finance"
})

before = len(df)
df = df.drop_duplicates()
records_removed = before - len(df)

df["Mark_Status"] = "Entered"

df.loc[df["Marks"].isna(), "Mark_Status"] = "Not Entered"
df.loc[df["Marks"].astype(str).str.lower().eq("absent"), "Mark_Status"] = "Absent"

df["Marks"] = (
    df["Marks"]
    .astype(str)
    .str.replace(" marks", "", regex=False)
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

invalid_marks = df.loc[
    (df["Marks"] < 0) | (df["Marks"] > 100),
    ["Student_ID", "Student_Name", "Marks"]
]

df.loc[(df["Marks"] < 0) | (df["Marks"] > 100), "Marks"] = pd.NA
df.loc[df["Marks"].isna() & (df["Mark_Status"] == "Entered"), "Mark_Status"] = "Invalid"

print("\nCleaned DataFrame:")
print(df)

print("\nRecords removed:", records_removed)
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:")
print(invalid_marks)
```

Explanation:

The cleaning process standardizes department names, removes duplicate records, converts marks to numeric values, preserves the reason for missing marks, and identifies invalid marks instead of blindly deleting them.
